```python
def print_grid_stats(mouse_data):
    grid_cells = mouse_data[(mouse_data["grid_score_x"] > 0.3) & (mouse_data["grid_score_y"] > 0.3)]
    print(f"out of {len(mouse_data)}, there is {len(grid_cells)}")

    for session_id in np.unique(mouse_data["session_id"]):
        session_data = mouse_data[mouse_data["session_id"] == session_id]
        grid_cells = session_data[(session_data["grid_score_x"] > 0.3) & (session_data["grid_score_y"] > 0.3)]
        print(f"{session_id}, out of {len(session_data)}, there is {len(grid_cells)}")

def grid_summary(Mouse_of1_data, Mouse_of2_data):
    of1 = Mouse_of1_data[['session_id', 'cluster_id', 'shank_id', 'grid_score']]
    of2 = Mouse_of2_data[['session_id', 'cluster_id', 'shank_id', 'grid_score']]
    M_data = of1.merge(of2, on=["session_id","cluster_id", "shank_id"])
    print_grid_stats(M_data)
    plt.scatter(M_data["grid_score_x"], M_data["grid_score_y"])
    plt.axhline(0.3, color="red")
    plt.axvline(0.3, color="red")
```
